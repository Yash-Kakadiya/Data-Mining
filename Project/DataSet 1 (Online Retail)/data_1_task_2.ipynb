# 🧺 Online Retail Dataset (Apriori Preprocessing) - TASK 2
**Dataset**: https://www.kaggle.com/datasets/vijayuv/onlineretail

# Step 1: Load Data

```python
import pandas as pd
```

```python
df = pd.read_csv("../data/OnlineRetail.csv", encoding='latin1')
df
```

# Step 2: Drop Missing Invoices or Descriptions

```python
df.dropna(subset=['InvoiceNo', 'Description'], inplace=True)
df
```

# Step 3: Remove Canceled Invoices <BR> Start with letter 'c'

```python
df = df[~df['InvoiceNo'].astype(str).str.startswith('C')]
df
```

# Step 4: Keep Only Positive Quantities

```python
df = df[df['Quantity'] > 0]
df
```

# Step 5: Standardize Item Descriptions | Lower case and remove space

```python
df['Description'] = df['Description'].str.strip().str.lower()
df
```

# Step 6: Filter for a Single Country (e.g., United Kingdom)

```python
df_uk = df[df['Country'] == 'United Kingdom']
df_uk
```

# Step 7: Remove Duplicates

```python
df_uk.drop_duplicates(inplace=True)
df_uk
```

# Step 8: Add TotalPrice column , TotalPrice = Quantity *  UnitPrice

```python
df_uk['TotalPrice'] = df_uk['Quantity'] * df_uk['UnitPrice']
df_uk
```

# Step 9 : Check for Top 10 Selling Products

```python
top_selling_products = df_uk.groupby('Description')['Quantity'].sum().sort_values(ascending=False).head(10)
display(top_selling_products)
```

# Step 10 : Identify Most Active Customers

```python
most_active_customers = df_uk.groupby('CustomerID')['InvoiceNo'].nunique().sort_values(ascending=False)
display(most_active_customers)
```

# Step 11: Time-Based Analysis | Convert InvoiceDate to datetime and check popular months/hours.

```python
df_uk['InvoiceDate'] = pd.to_datetime(df_uk['InvoiceDate'])
df_uk
```

```python
import matplotlib.pyplot as plt

df_uk['month'] = df_uk['InvoiceDate'].dt.month
monthly_transactions = df_uk.groupby('month')['InvoiceNo'].nunique()

plt.figure(figsize=(10, 6))
monthly_transactions.plot(kind='bar')
plt.title('Number of Transactions per Month (United Kingdom)')
plt.xlabel('Month')
plt.ylabel('Number of Transactions')
plt.xticks(rotation=0)
plt.show()
```

```python
basket = (df_uk.groupby(['InvoiceNo', 'Description'])['Quantity']
          .sum().unstack().reset_index().fillna(0)
          .set_index('InvoiceNo'))
basket
```
